```python
import matplotlib.pyplot as plt 
import numpy as np 
```

### Scatter Plot

```python
x_data=np.random.random(50)*100
y_data=np.random.random(50)*100

plt.scatter(x_data,y_data,c='#00f',s=60,marker="o",alpha=0.6)
plt.show()
```

### Line Plot

```python
years=[2004+x for x in range(23)]
weights=np.random.randint(low=0,high=200,size=23)
plt.plot(years,weights,c='g',lw=5,linestyle='--')
plt.show()
```

## Bar Plot

```python
x=['java','python','C++','C#','C','HTML',"CSS",'Javascript']
y=np.random.randint(low=5,high=300,size=8)

plt.bar(x,y,facecolor='red',align='edge',width=0.5,edgecolor='orange',lw=6)

plt.show()
```

## Histograms

```python
ages=np.random.normal(loc=20,scale=1.5,size=1000)

plt.hist(ages,bins=20,cumulative=True)

plt.show()
```

## Pie Charts

```python
langs=["python","Rust","GO","SQL"]
votes=np.random.randint(low=5,high=100,size=4)
explodes=[0,0,0.3,0]

plt.pie(votes,
        labels=langs,
        explode=explodes,
        autopct='%1.1f%%',
        pctdistance=1.3,
        startangle=90)
plt.show()
```

## Boxplots

```python
heights=np.random.normal(loc=172,scale=8,size=300)
plt.boxplot(heights)
plt.show()
```

```python
first=np.linspace(0,10,25)
second=np.linspace(10,200,25)
third=np.linspace(200,210,25)
fourth=np.linspace(210,230,25)

data=np.concatenate((first,second,third,fourth))

plt.boxplot(data)
plt.show()
```

## Plot Customization

```python
years=[2020,2021,2022,2023,2024,2025,2026]
income=[62,54,33,64,24,63,86]

income_ticks=list(range(20,100,10))


plt.title("income of bill (in USD)".upper(),fontsize=30,fontname="Arial")
plt.xlabel("Year".capitalize())
plt.ylabel("yearly income (in USD)".upper())
plt.yticks(income_ticks,[f"${x}k USD" for x in income_ticks])

plt.plot(years,income,marker='o')
plt.grid(True,linestyle="--",alpha=0.6)
plt.show()
```

##  Multiple Figures

```python
x1,y1=np.random.random(100),np.random.random(100)
x2,y2=np.arange(100),np.random.random(100)

plt.figure(1)
plt.scatter(x1,y1)

plt.figure(2)
plt.plot(x2,y2)

plt.show()
```

## Subplots

```python
plt.style.use("default")
x=np.arange(1,101)

fig, axs = plt.subplots(2, 2, figsize=(10, 6))

axs[0,0].plot(x,np.sin(x))
axs[0,0].set_title("Sine Wave")

axs[0,1].plot(x,np.cos(x))
axs[0,1].set_title("Cosine Wave")

axs[1,0].plot(x,np.random.random(100))
axs[1,0].set_title("Random Function")

axs[1,1].plot(x,np.log(x))
axs[1,1].set_title("Log Function")

fig.suptitle("Four Plots")
plt.tight_layout()
plt.show()
# plt.savefig("Four plots",dpi=300)
```

## 3D Plotting

```python

ax=plt.axes(projection="3d")

x=np.random.random(100)
y=np.random.random(100)
z=np.random.random(100)

ax.scatter(x,y,z)
ax.set_title("3D Plot")

plt.show()
```

```python
ax=plt.axes(projection="3d")

x=np.arange(0,50,0.1)
y=np.sin(x)
z=np.cos(x)

ax.scatter(x,y,z)
ax.set_title("3D Plot")

plt.show()
```

```python
ax=plt.axes(projection="3d")

x=np.arange(-5,5,0.1)
y=np.arange(-5,5,0.1)

X,Y=np.meshgrid(x,y)

Z=np.sin(X)*np.cos(Y)

ax.plot_surface(X,Y,Z,cmap="Spectral")
ax.set_title("3D Plot")

plt.tight_layout()
plt.show()
```

```python
ax=plt.axes(projection="3d")

x=np.arange(-5,5,0.1)
y=np.arange(-5,5,0.1)

X,Y=np.meshgrid(x,y)

Z=np.sin(X)*np.cos(Y)

ax.plot_surface(X,Y,Z,cmap="Spectral")
ax.set_title("3D Plot")
ax.view_init(azim=0,elev=90)

plt.tight_layout()
plt.show()
```
